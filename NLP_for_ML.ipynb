#text cleaning
1. lower casing
2. remove pantualion

```python
from ast import Lambda
df['text'] = df['text'].apply(lambda x : x.lower() )
```

```python
import string

def remove_punc(txt):
  return txt.translate(str .maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt):
  # Iterate through the entire text and filter out digits, then return the full cleaned text
  return ''.join([char for char in txt if not char.isdigit()])
```

```python
df['text']= df['text'] .apply(remove_numbers)
```

```python
def remove_emojis(text):
  # Keep only ASCII characters to remove non-ASCII emojis/symbols
  return ''.join([char for char in text if char.isascii()])
```

```python
df['text']= df['text'] .apply(remove_emojis)
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if i not in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
# 1. Reload the dataset fresh
df = pd.read_csv('/content/drive/MyDrive/train.txt', sep=';', header=None, names=['text','emotion'])

# Define the correct mapping from the original unique text emotions
unique_emotions = ['sadness', 'anger', 'love', 'surprise', 'fear', 'joy']
emotion_numbers = {emo: i for i, emo in enumerate(unique_emotions)}

# Map original text values to numeric
df['emotion'] = df['emotion'].map(emotion_numbers)

# 2. Text Cleaning Pipeline
df['text'] = df['text'].apply(lambda x: x.lower())
df['text'] = df['text'].apply(remove_punc)
df['text'] = df['text'].apply(remove_numbers)
df['text'] = df['text'].apply(remove_emojis)
df['text'] = df['text'].apply(remove)

display(df.head())
```

```python
from sklearn.model_selection import train_test_split

```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['emotion'], test_size=0.2, random_state=42
)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

```

```python
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
```

```python
X_train_bow
```

```python
# Use transform on test data, NOT fit_transform
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
nb_model=MultinomialNB()
nb_model.fit(X_train_bow,y_train)
```

```python
tfidf_transformer = TfidfTransformer()
X_train_tfidf = tfidf_transformer.fit_transform(X_train_bow)
```

```python
# Use transform on test data, NOT fit_transform
X_test_tfidf = tfidf_transformer.transform(X_test_bow)
```

```python
# Retrain NB model on the correct training data
nb_model = MultinomialNB()
nb_model.fit(X_train_tfidf, y_train)

# Predict using correctly transformed test data
pred_nb = nb_model.predict(X_test_tfidf)
print(f"Accuracy Score: {accuracy_score(y_test, pred_nb):.4f}\n")
print(classification_report(y_test, pred_nb))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
lr.fit(X_train_tfidf,y_train)
```

```python
lr_pred=lr.predict(X_test_tfidf)
print(f"Accuracy Score: {accuracy_score(y_test, lr_pred):.4f}\n")
print(classification_report(y_test, lr_pred))
```
